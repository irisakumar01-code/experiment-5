```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('img.jpg')
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(6, 6))
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
averaging = cv2.blur(img, (15, 15))

plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title('Averaging Filter')
plt.axis('off')
plt.show()
```

```python
kernel = np.array([[1, 2, 1],
                   [2, 4, 2],
                   [1, 2, 1]], dtype=np.float32) / 16

weighted = cv2.filter2D(img, -1, kernel)

plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title('Weighted Averaging Filter')
plt.axis('off')
plt.show()
```

```python
gaussian = cv2.GaussianBlur(img, (15, 15), 0)

plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Filter')
plt.axis('off')
plt.show()
```

```python
median = cv2.medianBlur(img, 15)

plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title('Median Filter')
plt.axis('off')
plt.show()
```

```python
lap_kernel = np.array([[0, -1, 0],
                       [-1, 5, -1],
                       [0, -1, 0]])

laplacian_kernel = cv2.filter2D(img, -1, lap_kernel)

plt.imshow(cv2.cvtColor(laplacian_kernel, cv2.COLOR_BGR2RGB))
plt.title('Laplacian Sharpening')
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

laplacian = cv2.Laplacian(gray, cv2.CV_64F)

laplacian = cv2.convertScaleAbs(laplacian)

plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Operator')
plt.axis('off')
plt.show()
```

```python
plt.figure(figsize=(18, 12))

plt.subplot(2, 4, 1)
plt.imshow(img_rgb)
plt.title('Original Image')
plt.axis('off')

plt.subplot(2, 4, 2)
plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title('Averaging Filter')
plt.axis('off')

plt.subplot(2, 4, 3)
plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title('Weighted Averaging')
plt.axis('off')

plt.subplot(2, 4, 4)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title('Gaussian Filter')
plt.axis('off')

plt.subplot(2, 4, 5)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title('Median Filter')
plt.axis('off')

plt.subplot(2, 4, 6)
plt.imshow(cv2.cvtColor(laplacian_kernel, cv2.COLOR_BGR2RGB))
plt.title('Laplacian Kernel')
plt.axis('off')

plt.subplot(2, 4, 7)
plt.imshow(laplacian, cmap='gray')
plt.title('Laplacian Operator')
plt.axis('off')

plt.tight_layout()
plt.show()
```
